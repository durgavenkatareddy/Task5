```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_csv('TitanicDataset.csv')

# Initial look
df.head()
df.info()
df.describe()
df.isnull().sum()
```

```python
df.drop(columns='Cabin', inplace=True)
```

```python
df['Age'] = df['Age'].fillna(df['Age'].median())
```

```python
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
```

```python
df.head()
df.info()
df.describe()
df.isnull().sum()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Select numerical columns for correlation
numeric_cols = ['Survived', 'Pclass', 'Age', 'SibSp', 'Parch', 'Fare']
corr = df[numeric_cols].corr()

plt.figure(figsize=(10, 6))
sns.heatmap(corr, annot=True, cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Heatmap of Numeric Features')
plt.show()
```

```python
selected_features = ['Survived', 'Pclass', 'Age', 'Fare']
sns.pairplot(df[selected_features], hue='Survived', diag_kind='hist')
plt.suptitle('Pairplot of Selected Features by Survival', y=1.02)
plt.show()
```

```python
sns.countplot(x='Sex', hue='Survived', data=df)
sns.countplot(x='Pclass', hue='Survived', data=df)
```

```python
sns.histplot(data=df, x='Age', hue='Survived', bins=30, kde=True)
```

```python
sns.boxplot(x='Survived', y='Fare', data=df)
```

```python
df['FamilySize'] = df['SibSp'] + df['Parch']
sns.histplot(data=df, x='FamilySize', hue='Survived', bins=10, multiple='stack')
```

```python
sns.countplot(x='Embarked', hue='Survived', data=df)
```

```python
sns.histplot(data=df, x='Age', bins=30, kde=True)
plt.title('Distribution of Passenger Age')
plt.show()
```

```python
sns.histplot(data=df, x='Fare', bins=40, kde=True)
plt.title('Distribution of Fare Paid')
plt.show()
```

```python
sns.countplot(x='Survived', data=df)
plt.title('Survival Counts (0 = Died, 1 = Survived)')
plt.show()
```

```python
sns.boxplot(x='Survived', y='Age', data=df)
plt.title('Age Distribution by Survival')
plt.show()
sns.boxplot(x='Pclass', y='Fare', data=df)
plt.title('Fare Distribution Across Passenger Classes')
plt.show()
sns.boxplot(x='Pclass', y='Age', data=df)
plt.title('Age Distribution Across Passenger Classes')
plt.show()
```

```python
sns.scatterplot(x='Age', y='Fare', hue='Survived', data=df)
plt.title('Fare vs Age Colored by Survival')
plt.show()
sns.scatterplot(x='Pclass', y='Fare', hue='Survived', data=df)
plt.title('Pclass vs Fare Colored by Survival')
plt.show()
```
